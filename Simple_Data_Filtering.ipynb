# Simple Data Filtering 📊

## Task 8 - Level 1 - Day 8

This project creates student records and filters students based on marks, age, department, and multiple conditions.


````python
students = [
    {"id":"S001","name":"Rahul","age":20,"marks":85,"department":"AI & ML"},
    {"id":"S002","name":"Priya","age":21,"marks":91,"department":"Data Science"},
    {"id":"S003","name":"Aman","age":19,"marks":72,"department":"Computer Science"},
    {"id":"S004","name":"Neha","age":20,"marks":88,"department":"AI & ML"},
    {"id":"S005","name":"Riya","age":22,"marks":95,"department":"Data Science"},
    {"id":"S006","name":"Arjun","age":21,"marks":67,"department":"Computer Science"}
]
print("Total Students:", len(students))
````

## 1. Display All Students

The following code displays all student records.


````python
for student in students:
    print(student)
````

## 2. Filter Students Based on Marks

Select students with marks greater than or equal to 85.


````python
marks_filtered = [student for student in students if student["marks"] >= 85]
for student in marks_filtered:
    print(student)
````

## 3. Filter Students Based on Age

Select students whose age is less than or equal to 20.


````python
age_filtered = [student for student in students if student["age"] <= 20]
for student in age_filtered:
    print(student)
````

## 4. Filter Students Based on Department

Select students belonging to the AI & ML department.


````python
department_filtered = [
    student for student in students
    if student["department"] == "AI & ML"
]
for student in department_filtered:
    print(student)
````

## 5. Multiple Filtering Conditions

Select students with marks >= 80 and age <= 20.


````python
multiple_filtered = [
    student for student in students
    if student["marks"] >= 80 and student["age"] <= 20
]
for student in multiple_filtered:
    print(student)
````

## 📊 Filtering Results

| Filter | Condition |
|---|---|
| Marks | Marks >= 85 |
| Age | Age <= 20 |
| Department | AI & ML |
| Multiple Conditions | Marks >= 80 and Age <= 20 |


````python
print("Marks Filter:", len(marks_filtered))
print("Age Filter:", len(age_filtered))
print("Department Filter:", len(department_filtered))
print("Multiple Condition Filter:", len(multiple_filtered))
````

### 3. What is the difference between filtering and sorting?
Filtering selects records that meet a condition, while sorting changes the order of records.
